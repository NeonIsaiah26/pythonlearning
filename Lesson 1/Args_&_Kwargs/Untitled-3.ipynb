```python
def order_pizza(size, *toppings, **details):
    print(f"Ordered a {size} pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")
    print("Additional details:")
    print(f"Delivery Time: {details.get('delivery_time')}")
    print(f"Address: {details.get('address')}")
```

```python
order_pizza("Large", 'pepperoni', "mushrooms", "extra cheese", address="123 Main St", delivery_time="30 minutes")
```
